```python
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image

# Carrega modelo e processador (uma vez só)
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")

def descrever_imagem(caminho_imagem, prompt=None, max_length=50, num_beams=5):
    # Entrada
    imagem = Image.open(caminho_imagem).convert("RGB")

    # Processamento
    # Se um prompt for fornecido, passe-o para o processador
    if prompt:
        inputs = processor(imagem, text=prompt, return_tensors="pt")
    else:
        inputs = processor(imagem, return_tensors="pt")

    # Gerar output com max_length e num_beams para uma descrição mais detalhada
    # max_length controla o comprimento máximo da descrição.
    # num_beams aumenta a diversidade e qualidade das descrições geradas.
    output = model.generate(**inputs, max_length=max_length, num_beams=num_beams)

    # Saída
    descricao = processor.decode(output[0], skip_special_tokens=True)

    return descricao
```

```python
descricao = descrever_imagem("WhatsApp Image 2026-04-20 at 14.42.53.jpeg")
print(descricao)
```
