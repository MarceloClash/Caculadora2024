```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import numpy as np
import matplotlib.pyplot as plt
```

```python
import pandas as pd

# Dados de exemplo

data = {
    'caracteristica1': [1.2, 2.3, 3.4],
    'caracteristica2': [2.1, 3.2, 4.3],
    'caracteristica3': [3.2, 4.3, 5.4],
    'caracteristica4': [7.8, 8.9, 9.0],
    'caracteristica5': [8.9, 9.0, 10.1]
}
df = pd.DataFrame(data)
```

```python
# Padronizando os dados

scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
```

```python
# Aplicando o PCA

pca = PCA(n_components = 2)
principalComponents = pca.fit_transform(df_scaled)
```

```python
# Criando um dataframe com os componetes principais

df_pca = pd.DataFrame(data = principalComponents, columns = ['principal component 1', 'principal component 2'])
```

```python
# Visualizando os componetes principais

plt.scatter(df_pca['principal component 1'], df_pca['principal component 2'])
plt.title('PCA dos dados')
plt.xlabel('Principal component 1')
plt.ylabel('Principal component 2')
plt.show()
```

```python
# Exibindo a variância por cada componente

print('Variância por cada componente:', pca.explained_variance_ratio_)
```
